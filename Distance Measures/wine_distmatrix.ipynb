```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from scipy.spatial.distance import euclidean, cityblock, minkowski

```

```python
data = load_wine(as_frame=True)

df = data.frame
```

```python
subset = df.iloc[:3, :4]

print("Subset of data:\n")
print(subset)
```

```python
scaler = StandardScaler()
subset_scaled = scaler.fit_transform(subset)
```

```python
A = subset_scaled[0]
B = subset_scaled[1]
C = subset_scaled[2]
```

```python

# Compute distances
print("\nDistances between Sample 0 and Sample 1:")
print("Euclidean Distance:", euclidean(A, B))
print("Manhattan Distance:", cityblock(A, B))
print("Minkowski Distance (p=3):", minkowski(A, B, p=3))

print("\nDistances between Sample 0 and Sample 2:")
print("Euclidean Distance:", euclidean(A, C))
print("Manhattan Distance:", cityblock(A, C))
print("Minkowski Distance (p=3):", minkowski(A, C, p=3))

print("\nDistances between Sample 1 and Sample 2:")
print("Euclidean Distance:", euclidean(B, C))
print("Manhattan Distance:", cityblock(B, C))
print("Minkowski Distance (p=3):", minkowski(B, C, p=3))
```
